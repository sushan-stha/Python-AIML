Question 3: Indexing a 4D Array

```python
import numpy as np
tensor_box = np.arange(1, 49).reshape(2, 2, 3, 4)
```

Print the first 3D block

```python
print("1.")
print(tensor_box[0])
```

```python
print("\n2.")
print(tensor_box[0, 1])
```

```python
print("\n3.")
print(tensor_box[1, 1, 2])
```

```python
print("\n4.")
print(tensor_box[0, :, :, -1])
```

```python
print("\n5.")
print(tensor_box[1, 1, 2, 3])
```

```python
print("\n6.")
print(tensor_box[1, 0, :, :2])
```

Question 4: Advanced Slicing on a 4D Array

```python
matrix_hub = np.arange(1, 97).reshape(2, 3, 4, 4)
```

```python
print("1.")
print(matrix_hub[1])
```

```python
print("\n2.")
print(matrix_hub[:, :, :2, :])
```

```python
print("\n3.")
print(matrix_hub[:, :, :, 2:])
```

```python
print("\n4.")
print(matrix_hub[:, :, ::2, :])
```

```python
print("\n5.")
print(matrix_hub[:, :, 1:3, 1:3])
```
