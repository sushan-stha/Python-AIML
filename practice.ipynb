```python
import pandas as pd
import numpy as np

df = pd.read_csv("day_10_dataset.csv")

print(df.shape)
print(df.head())
print(df.info())
```

Convert numeric columns

```python
num_cols = [
    'Customer_Age', 'Quantity', 'Unit_Price_NPR',
    'Shipping_Fee_NPR', 'Cost_NPR', 'Rating', 'Delivery_Days'
]

for col in num_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

Cleaning discount 

```python
df['Discount'] = df['Discount'].astype(str).str.strip()

df['Discount'] = pd.to_numeric(
    df['Discount'].str.replace('%', '', regex=False),
    errors='coerce'
)

df.loc[df['Discount'] > 1, 'Discount'] = df.loc[df['Discount'] > 1, 'Discount'] / 100

df['Discount'] = df['Discount'].fillna(0)
```

Cleaning dates

```python
df['Order_Date'] = pd.to_datetime(
    df['Order_Date'],
    errors='coerce',
    format='mixed'
)

df['Delivery_Date'] = pd.to_datetime(
    df['Delivery_Date'],
    errors='coerce',
    format='mixed'
)
```

Removing dublicate orders

```python
df = df.drop_duplicates(subset='Order_ID')
```

Cleaning category

```python
df['Product_Category'] = df['Product_Category'].str.strip().str.lower()

df['Product_Category'] = df['Product_Category'].replace({
    'fashon': 'fashion',
    'electronic': 'electronics',
    'book': 'books',
    'books': 'books',
    'grocery': 'groceries',
    'home and kitchen': 'home & kitchen'
})

df['Product_Category'] = df['Product_Category'].str.title()
```

Cleaning order status

```python
df['Order_Status'] = df['Order_Status'].str.strip().str.lower()

df['Order_Status'] = df['Order_Status'].replace({
    'delivered': 'Delivered',
    'cancelled': 'Cancelled',
    'canceled': 'Cancelled',
    'returned': 'Returned',
    'return': 'Returned',
    'failed': 'Failed',
    'payment failed': 'Payment Failed',
    'pending': 'Pending'
})
```

Cleaning sales channel

```python
df['Sales_Channel'] = df['Sales_Channel'].str.strip().str.lower()

df['Sales_Channel'] = df['Sales_Channel'].replace({
    'daraz': 'Daraz Marketplace',
    'daraz marketplace': 'Daraz Marketplace',
    'website': 'Website',
    'web': 'Website',
    'app': 'Mobile App',
    'mobile app': 'Mobile App',
    'instagram': 'Instagram',
    'insta': 'Instagram'
})
```

Cleaning payment methods

```python
df['Payment_Method'] = df['Payment_Method'].str.strip().str.lower()

df['Payment_Method'] = df['Payment_Method'].replace({
    'khalti': 'Khalti',
    'bank transfer': 'Bank Transfer',
    'esewa': 'eSewa',
    'e-sewa': 'eSewa',
    'card': 'Card',
    'credit card': 'Card',
    'debit card': 'Card',
    'cod': 'Cash on Delivery',
    'cash on delivery': 'Cash on Delivery',
    'cash-on-delivery': 'Cash on Delivery'
})
```

Cleaning Return flags

```python
df['Return_Flag'] = df['Return_Flag'].str.strip().str.lower()

df['Return_Flag'] = df['Return_Flag'].replace({
    'yes': 'Yes',
    'y': 'Yes',
    'no': 'No',
    'n': 'No'
})
```
